```python
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime,timedelta #datetime – Date aur time handle karta hai.
# Do dates ke darmiyan time difference ya date mein days add/subtract karta hai.
```

```python
# generate date list for an year
start_date = datetime(2024,1,1)
end_date = datetime(2024,12,31)
num_days = (end_date - start_date).days #(Ye timedelta object mein se sirf days nikalta hai.)
# (Jab hum Python mein do dates ko subtract karte hain, to Python automatically unke darmiyan ka time difference calculate karta hai.)
print(start_date,num_days)
```

```python
date_list =[]
for i in range(num_days):
  # append() list mein new item add karta hai.
  # start_date.date() sirf date return karta hai, time nahi.
    date_list.append(start_date.date())
    # Iska matlab hai har iteration ke baad date mein 1 day add karna.
    start_date += timedelta(days=1)

date_list
```

```python
# random stock prices generate karne ke liye use ho rahi hai
stock_price_list = (np.random.randn(num_days)).tolist() # (.tolist()NumPy array ko Python list banana)
#np.random (Ye NumPy ka random number generation module hai.)
# Iske andar different functions hote hain jo different types ke random numbers generate karte hain.
# randn (normally distributed random numbers generate karega.)
```

```python
print(np.random.randn(5))
print(np.random.randn(5))
```

```python
np.random.seed(0)
print(np.random.randn(5))

np.random.seed(0)
print(np.random.randn(5))
```

```python
np.random.seed(0)
# np.random.seed(0) ka matlab hai ke har baar code run karne par same random stock prices generate honge.
# cumsum() – Numbers ko cumulative add karta hai.
# Cumulative Add ka matlab hai har naye number ko us se pehle wale tamam numbers ke total mein add karna.
stock_price_list = (np.random.randn(num_days).cumsum()* 50).tolist()
stock_price_list
```

```python
plt.Figure(figsize=(15,6))
# Graph ki size set karta hai:(15 = Width)(6 = Height)
plt.plot(date_list,stock_price_list,color ='red',label='price')
plt.title("Stock price for 1 year")
plt.xlabel("date")
plt.ylabel("stock prices")
plt.legend()
plt.grid(True)
# Graph ke background mein horizontal aur vertical lines show karta hai, jisse values read karna easy hota hai.
plt.tight_layout()
# Graph ke title aur labels ko properly adjust karta hai taake text cut na ho.
plt.show()
```

```python
plt.Figure(figsize=(15,6))
plt.plot(date_list,stock_price_list,color ='green',marker = '*',label='price')
plt.title("Stock price for 1 year")
plt.xlabel("date")
plt.ylabel("stock prices")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.Figure(figsize=(15,6))
plt.plot(date_list,stock_price_list,color ='green',linestyle = '--',label='price')
plt.title("Stock price for 1 year")
plt.xlabel("date")
plt.ylabel("stock prices")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
