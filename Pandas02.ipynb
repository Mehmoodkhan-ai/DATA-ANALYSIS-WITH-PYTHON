```python
titanic['Discount Fare'] = titanic['Fare'] / 10
titanic
```

```python
# if age < 10 then 25%
# if age 20 & 50 then 10 %
# if age > 50 then 50%
```

```python
def apply_discount(age):
    if age < 10 :
        return 0.25
    elif 20 <= age <= 50:
        return 0.1
    elif age > 50:
        return 0.5

titanic['Discount'] = titanic['Age'].apply(apply_discount)
titanic['Discount Fare'] = titanic['Fare'] * (1-titanic['Discount']) # 1 - 0.1 == 0.9
titanic[['Name','Fare','Age','Discount','Discount Fare']]
```

Pandas mein Quantile ek statistical concept hai jo data ko equal parts mein divide karta hai aur batata hai ke kisi particular percentage par kaunsi value aati hai.

Simple words mein: Quantile batata hai ke data ki kitni percentage values kisi specific value ke barabar ya us se neeche hain.

```python
titanic['Age'].quantile() #<--Return values at the given quantile over requested axis.
# By default 0.5
```

```python
titanic['Age'].quantile(0.25) #<--Return values at the given quantile over requested axis.
# By default 0.5
```

```python
titanic['Age'].quantile(0.3) #<--Return values at the given quantile over requested axis.
```
