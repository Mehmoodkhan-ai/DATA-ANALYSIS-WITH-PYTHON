# ***Matplotlib***
Python ki ek popular aur powerful data visualization library hai. Iska use data ko graphs, charts aur different types of plots ki form mein visually represent karne ke liye kiya jata hai.

Matplotlib ki madad se hum complex data ko easily understand kar sakte hain. Is library ke through hum Line Graph, Bar Chart, Pie Chart, Histogram aur Scatter Plot create kar sakte hain.

Matplotlib data analysis aur data science mein bohat important role play karti hai, kyun ke ye numerical data ko visual form mein convert karti hai

```python
# Line Chart

year =  [2020,2021,2022,2023,2024,2025]
sales = [100,105,120,130,270,70]

# Ye function years aur sales ke darmiyan line chart create karta hai.
plt.plot(year,sales)

plt.show()
```

```python
year =  [2020,2021,2022,2023,2024,2025]
sales = [100,105,120,130,270,70]

# making a plot
plt.plot(year,sales)
plt.title("sales Analysis")
# show a plot
plt.show()
```

```python
year =  [2020,2021,2022,2023,2024,2025]
sales = [100,105,120,130,270,70]

# making a plot
plt.plot(year,sales)
plt.title("sales Analysis")
plt.xlabel("Time")
plt.ylabel("Sales")

# show a plot
plt.show()
```

```python
year =  ['jan','feb','mar','apr']
sales = [100,105,120,130]

# making a plot
plt.bar(year,sales,color='black')
plt.title("sales Analysis")
plt.xlabel("Months")
plt.ylabel("Sales")

# show a plot
plt.show()
```
