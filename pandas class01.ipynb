## Dictionaries in Python

```python
# Dictionary data ko Key-Value pairs ki form mein store karti hai.
countries = {
    "Pak": 45.55,
    "Afg": 55.22,
    "India": 10.33,
    "China": 68.88,
    "Iraq": 71.55,
    "Japan": 96.22
}
```

```python
# China ki population nikalne ke liye
print(countries["China"])
# Ab index find karne ki zaroorat nahi hai.
# "China" = Key
# 68.88 = Value
```

Dictionary mein different types ka data store kar sakte hain.

```python
student = {
    "name": "Ali",
    "age": 20,
    "marks": 85.5,
    "passed": True
}
```

- Immutable ka matlab: Aisi value jise create hone ke baad modify nahi kiya ja sakta.
- Common examples:
- String
- Integer
- Float
- Tuple (agar uske elements bhi hashable hon)

```python
data = {
    "name": "Ali",
    10: "Number",
    (1, 2): "Tuple"
}
```

```python
data = {
    [1, 2]: "List"
}
# yah error dega kyunki list mutable hoti hai aur dictionary ke keys immutable hone chahiye.
# List = Mutable
# Dictionary = Mutable
# Tuple = Immutable
# String = Immutable
```

## CREATING DATAFRAME
- Dictionary ko Pandas ke DataFrame mein convert kar rahe hain.

```python
bricks = {"country": ['Brazil', 'Russia', 'India', 'China', 'South Africa'],
         "capital": ['Brasilia', 'Moscow', 'New Delhi', 'Beijing', 'Pretoria'],
         "area": [8.516, 17.100, 3.286, 9.597, 1.221],
         "population": [200.40, 143.50, 1252.00, 1357.00, 52.98]}
bricks
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(bricks)
df
```

```python
df.shape
```

```python
# set_index('country'): Country column ko index bana deta hai.
# inplace=True: Original DataFrame (df) mein hi change karta hai.
df.set_index('country',inplace=True)  
df
```

- CSV file se directly DataFrame create kar rahe hain.
- pd.read_csv() → CSV file ko read karke DataFrame banata hai.
- index_col=0 → CSV ka pehla column DataFrame ka index ban jayega.
- df → Created DataFrame ko store karta hai.

```python
df = pd.read_csv('./dataset/bricks.csv', index_col=0)
df
```

```python
# check karna ke df kis data type ka object hai.
type(df)
# Yani df ek Pandas DataFrame hai.
```

```python
#Ye check karta hai ke df ke country column ka data type kya hai. 
type(df.country)
#  DataFrame ka single column ek Series hota hai.
```

```python
# DataFrame se BR index wali complete row select karna.
# loc → Label ke through data access karta hai.
# 'BR' → Index label hai.

df.loc['BR']
```

```python
# df.loc['BR'] → Single row.
# df.loc[['BR', 'RU']] → Multiple rows.
df.loc[['BR','RU']]
```

```python
df.loc[0]
```

```python
df.iloc[0]
# iloc → Integer position ke through data access karta hai.
# 0 → Pehli row.
# Output → Brazil ki complete information
```

```python
df.iloc[[0,1,2]]
```

```python
df.iloc[:3]
#  Python slicing mein ending position include nahi hoti.
```

```python
#rows aur columns dono select kar rahe hain.
# 1:3 → Row position 1 se 2 tak (2 rows).
# :2 → Column position 0 se 1 tak (2 columns).
df.iloc[1:3, :2]
```

```python
# Comma se pehle = Rows
# Comma ke baad = Columns
# Ending position include nahi hoti.
df.iloc[1:3, 2:4]
```

```python
df.loc[['BR','RU'],['country','capital']]
```
